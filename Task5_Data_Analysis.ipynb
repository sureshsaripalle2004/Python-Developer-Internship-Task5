Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

Read the CSV File

```python
df = pd.read_csv("Sale Report.csv")
```

Check Missing Values

```python
df.isnull().sum()
```

Total Stock by Color

```python
color_stock = df.groupby("Color")["Stock"].sum()
print(color_stock)
```

Bar Chart - Stock by Color

```python
color_stock.plot(kind="bar", figsize=(8,5))
plt.title("Total Stock by Color")
plt.xlabel("Color")
plt.ylabel("Stock")
plt.tight_layout()
plt.savefig("stock_by_color.png")
plt.show()
```

Total Stock by Size

```python
size_stock = df.groupby("Size")["Stock"].sum()
print(size_stock)
```

Bar Chart - Stock by Size

```python
size_stock.plot(kind="bar", figsize=(8,5))
plt.title("Total Stock by Size")
plt.xlabel("Size")
plt.ylabel("Stock")
plt.tight_layout()
plt.savefig("stock_by_size.png")
plt.show()
```

Total Stock by Category

```python
category_stock = df.groupby("Category")["Stock"].sum()
print(category_stock)
```

Bar Chart - Stock by Category

```python
category_stock.plot(kind="bar", figsize=(8,5))
plt.title("Total Stock by Category")
plt.xlabel("Category")
plt.ylabel("Stock")
plt.tight_layout()
plt.savefig("stock_by_category.png")
plt.show()
```

Highest Stock Item

Lowest Stock Item

```python
df[df["Stock"] == df["Stock"].min()]
```

Filter Rows (Stock > 10)

```python
df[df["Stock"] > 10]
```

```python
df.sort_values(by="Stock", ascending=False)
```
